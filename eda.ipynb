# Complete EDA Notebook

This notebook is designed for a Kaggle-style CSV dataset.

How to use it:
1. Put your Kaggle CSV file inside the `data` folder.
2. Run the cells from top to bottom.
3. Charts will be shown inline and also saved to the `output` folder.

```python
from pathlib import Path

import pandas as pd
from IPython.display import display

from eda import (
    categorical_summary,
    column_summary,
    date_summary,
    get_column_groups,
    grouped_numeric_summary,
    load_data,
    missing_summary,
    numeric_summary,
    outlier_summary,
    plot_boxplots,
    plot_categorical_counts,
    plot_correlation_heatmap,
    plot_missingness,
    plot_numeric_distributions,
    prepare_dates,
    profile_overview,
    set_plot_style,
)

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 200)

try:
    set_plot_style()
except ImportError as exc:
    print(exc)
```

## Load The Dataset

```python
csv_path, df = load_data()
df, parsed_dates = prepare_dates(df)

print(f"Loaded file: {csv_path}")
print(f"Shape: {df.shape[0]} rows x {df.shape[1]} columns")
if parsed_dates:
    print("Parsed date columns:", ", ".join(parsed_dates))

df.head()
```

## Quick Overview

```python
overview = profile_overview(df)
overview
```

```python
display(column_summary(df))
```

## Missing Values And Duplicates

```python
print("Duplicate rows:", int(df.duplicated().sum()))
display(missing_summary(df))
```

## Numeric, Categorical, And Date Columns

```python
numeric_cols, categorical_cols = get_column_groups(df)

print("Numeric columns:", numeric_cols)
print("Categorical columns:", categorical_cols)
display(date_summary(df))
```

## Univariate Analysis

```python
display(numeric_summary(df))
display(outlier_summary(df))
display(categorical_summary(df))
```

## Visual Analysis

```python
plot_missingness(df)
plot_numeric_distributions(df)
plot_boxplots(df)
plot_correlation_heatmap(df)
plot_categorical_counts(df)
```

## Grouped Analysis

```python
grouped_tables = grouped_numeric_summary(df)
for name, table in grouped_tables.items():
    print(f"\nAverage numeric values grouped by: {name}")
    display(table)
```
